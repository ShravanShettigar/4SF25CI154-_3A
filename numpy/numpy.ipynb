```python

import numpy as np

arr = np.random.randint(1, 21, (5, 5))

print("Original Array:")
print(arr)

arr[:, 2] = 1

print("Modified Array:")
print(arr)
```

```python

import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Original Array:")
print(arr)

np.fill_diagonal(arr, 0)

print("Modified Array:")
print(arr)
```

```python

import numpy as np

arr = np.random.randint(1, 20, (5, 5))

print("Original Array:")
print(arr)

border = np.concatenate((
    arr[0, :],
    arr[-1, :],
    arr[1:-1, 0],
    arr[1:-1, -1]
))

print("Border Elements:")
print(border)
```

```python

import numpy as np

arr = np.arange(1, 17).reshape(4, 4)

print("Array:")
print(arr)

row_sum = np.sum(arr, axis=1)
col_sum = np.sum(arr, axis=0)

print("Row-wise Sum:", row_sum)
print("Column-wise Sum:", col_sum)
```

```python

import numpy as np

arr = np.arange(1, 10).reshape(3, 3)

print("Original Array:")
print(arr)

a = arr.reshape(1, 9)
print("Shape (1, 9):")
print(a)

b = arr.reshape(9, 1)
print("Shape (9, 1):")
print(b)
```

```python

import numpy as np

data = np.array([
    ("Rahul", 22, 65.5),
    ("Amit", 20, 60.2),
    ("Kiran", 24, 70.5),
    ("Arun", 19, 55.5)
], dtype=[
    ("name", "U20"),
    ("age", "i4"),
    ("weight", "f4")
])

print("Original Array:")
print(data)

sorted_data = np.sort(data, order="age")

print("Sorted by Age:")
print(sorted_data)
```
