```python

def count_words():
    with open("document.txt", "r") as file:
        data = file.read()

    words = data.split()
    return len(words)


print("Total words:", count_words())
```

```python

def replace_word():
    with open("document.txt", "r") as file:
        data = file.read()

    old_word = input("Enter word to find: ")
    new_word = input("Enter replacement word: ")

    data = data.replace(old_word, new_word)

    with open("document.txt", "w") as file:
        file.write(data)

    print("Replacement completed!")


replace_word()
```

```python

def reverse_lines():
    with open("document.txt", "r") as file:
        for line in file:
            print(line.rstrip()[::-1])


reverse_lines()
```

```python

def count_stats():
    with open("stats.txt", "r") as file:
        data = file.read()

    lines = len(data.splitlines())
    words = len(data.split())
    characters = len(data)

    print("Lines:", lines)
    print("Words:", words)
    print("Characters:", characters)


count_stats()
```
