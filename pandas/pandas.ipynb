```python

import pandas as pd
import numpy as np

df = pd.DataFrame(
    np.random.randint(1, 10, (4, 3)),
    columns=['A', 'B', 'C']
)

print("DataFrame:")
print(df)

print("\nRow-wise Sum:")
print(df.sum(axis=1))

print("\nColumn-wise Sum:")
print(df.sum(axis=0))
```

```python

import pandas as pd
import numpy as np

df = pd.DataFrame(
    np.random.randint(1, 20, (5, 3)),
    columns=['A', 'B', 'C']
)

df.iloc[1, 0] = np.nan
df.iloc[2, 1] = np.nan
df.iloc[3, 2] = np.nan

print("Original DataFrame:")
print(df)

df = df.fillna(df.mean())

print("\nAfter Filling NaN:")
print(df)
```

```python

import pandas as pd
import numpy as np

df = pd.DataFrame(
    np.random.randint(1, 20, (6, 4)),
    columns=['A', 'B', 'C', 'D']
)

df.iloc[1, 2] = np.nan
df.iloc[3, 0] = np.nan
df.iloc[5, 1] = np.nan

print("Original DataFrame:")
print(df)

df = df.dropna()

print("\nAfter Dropping NaN Rows:")
print(df)
```

```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(10, 100, 10)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("\nGrouped Data:")
print(result)
```

```python

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Mouse', 'Keyboard', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics',
                 'Accessories', 'Accessories', 'Electronics'],
    'Sales': np.random.randint(100, 1000, 6)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Sales'].sum()

print("\nTotal Sales by Category:")
print(result)
```

```python

import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Amit', 'Rahul', 'Kiran', 'Arun']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [85, 90, 75, 95]
})

print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)

result = pd.merge(df1, df2, on='ID')

print("\nMerged DataFrame:")
print(result)
```

```python

import pandas as pd

index = pd.MultiIndex.from_tuples([
    ('A', 1),
    ('A', 2),
    ('B', 1),
    ('B', 2),
    ('C', 1),
    ('C', 2)
], names=['Category', 'Number'])

df = pd.DataFrame({
    'Value': [10, 20, 30, 40, 50, 60]
}, index=index)

print("MultiIndex DataFrame:")
print(df)

print("\nAccess Category A:")
print(df.loc['A'])

print("\nAccess Category B, Number 2:")
print(df.loc[('B', 2)])

print("\nSlicing:")
print(df.loc['A':'B'])
```

```python

import pandas as pd
import numpy as np

df = pd.DataFrame(
    np.random.randint(1, 10, (5, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

def double(x):
    return x * 2

result = df.map(double)

print("\nAfter Doubling:")
print(result)
```

```python

import pandas as pd
import numpy as np

df = pd.DataFrame(
    np.random.randint(1, 10, (6, 3)),
    columns=['A', 'B', 'C']
)

print("Original DataFrame:")
print(df)

df['Total'] = df.apply(
    lambda row: row['A'] + row['B'] + row['C'],
    axis=1
)

print("\nAfter Applying Lambda:")
print(df)
```
